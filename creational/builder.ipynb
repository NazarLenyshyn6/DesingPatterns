### 1. **The problem builder solves (before the pattern)**
Imagine this object:

```python
from dataclasses import dataclass

@dataclass
class ModelConfig:
    model_id: str
    temperature: float
    max_tokens: int
    retries: int 
    timeout: int
    log_tokens: bool
    cache_enabled: bool
    system_prompt: str
    safety_rules: list
    
ModelConfig(
    model_id="gpt-4.1-mini",
    temperature=0.2,
    max_tokens=4096,
    retries=3,
    timeout=10,
    log_tokens=True,
    cache_enabled=True,
    system_prompt="...",
    safety_rules=[]
)
```

### What goes wrong over time

1. Constructo becomes huge
2. Parameters are optional but interdependent
3. Order of steps matters
4. Some combinations are invalid
5. different contexts need different subsets

You end up with:
- unreadable constructos
- many Nones
- runtime errors
- duplicated setup logic

**Factory does not solve this**, because Factory asnwers:
- *Which object should I create?*

But here the question is:
- *How do I safely and flexibly construct one complex object?*

That's the **Builder problem**

### 2. **What Builder is**

A **Builder** is:
- *A separate object that constructs another object step-by-step, allowing different configuratoins withought exploding constructors*

**Key ideas**:
- Constructoin is **sequential**
- Object is created only at the end
- Same object type, different configuratoins

```python
from typing import Optional
from dataclasses import dataclass

@dataclass
class Report:
    title: Optional[str] = None
    body: Optional[str] = None
    footer: Optional[str] = None
    logo: Optional[str] = None
    watermark: Optional[str] = None
    
    def __str__(self) -> str:
        return (
            f"Title: {self.title}\n"
            f"Body: {self.body}\n"
            f"Footer: {self.footer}\n"
        )
    
    
class ReportBuilder:
    def __init__(self):
        self._report = Report()
        
    def add_title(self, value: str):
        self._report.title = value
        return self
    
    def add_body(self, value: str):
        self._report.body = value
        return self
    
    def add_footer(self, value: str):
        self._report.footer = value
        return self
        
    def build(self) -> Report:
        if self._report.title is None or self._report.body is None:
            raise ValueError("Title and bode are required")
        return self._report
    
report = (
    ReportBuilder()
    .add_title("Monthly Stats")
    .add_body("Some content")
    .add_footer("Confidential")
    .build()
)
print(report)
```
